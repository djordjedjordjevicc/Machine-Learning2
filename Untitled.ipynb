**Otkrivanje outliear-a**

```python
# Izdvajanje numeričkih kolona
numerical_columns = dataset.select_dtypes(include=['int64', 'float64'])

# Računanje Q1 (25%) i Q3 (75%) za svaki numerički atribut
Q1 = numerical_columns.quantile(0.25)
Q3 = numerical_columns.quantile(0.75)
IQR = Q3 - Q1

# Identifikacija outlier-a: podaci koji su van IQR opsega
outlier_condition = (numerical_columns < (Q1 - 2.0 * IQR)) | (numerical_columns > (Q3 + 2.0 * IQR))
outliers = numerical_columns[outlier_condition.any(axis=1)]

# Broj outlier-a
num_outliers = len(outliers)
print(f"Broj outlier-a: {num_outliers}")
```

**Elbow metoda**

```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
kmeans_kwargs = {
     "init": "random",
     "n_init": 10,
     "max_iter": 300,
     "random_state": 42,
 }

sse = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, init="k-means++", n_init=10, max_iter=300, random_state=42)
    kmeans.fit(scaled_df)
    sse.append(kmeans.inertia_)
```

```python
plt.style.use("fivethirtyeight")
plt.plot(range(1, 11), sse)
plt.xticks(range(1, 11))
plt.xlabel("Number of Clusters")
plt.ylabel("SSE")
plt.show()
```

```python
from kneed import KneeLocator

kl = KneeLocator(range(1, 11), sse, curve="convex", direction="decreasing")
kl.elbow
```

**Optimalan broj klastera na osnovu elbow je 3**

**Silhouette koeficijent**

```python
from sklearn.metrics import silhouette_score

silhouette_coefficients = []

for k in range(2, 10):
    kmeans = KMeans(n_clusters=k,  random_state=42)
    kmeans.fit(scaled_df)
    score = silhouette_score(scaled_df, kmeans.labels_)
    silhouette_coefficients.append(score)
```

```python
print(silhouette_coefficients)
```

```python
plt.style.use("fivethirtyeight")
plt.plot(range(2, 10), silhouette_coefficients)
plt.xticks(range(2, 10))
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Coefficient")
plt.show()
```

**Yellowbrick - izbor optimalnog broja klastera**

```python
from sklearn.cluster import KMeans
from yellowbrick.cluster import KElbowVisualizer

def elbowDist(df_bez_dropout_scaled: pd.DataFrame):
    elbow = KElbowVisualizer(KMeans(), k=(2,12))
    elbow.fit(df_bez_dropout_scaled)
    elbow.show()
```

```python
def elbowCH(df_bez_dropout_scaled: pd.DataFrame):
    elbow = KElbowVisualizer(KMeans(), k=(2,12), metric='calinski_harabasz')
    elbow.fit(df_bez_dropout_scaled)
    elbow.show()
```

**Gausian mixture**

```python
from sklearn.mixture import GaussianMixture

n_clusters_range = range(1, 6)
bics = []
aics = []

for n_clusters in n_clusters_range:
    gmm = GaussianMixture(n_components=n_clusters, random_state=42)
    gmm.fit(scaled_df)
    bics.append(gmm.bic(scaled_df))
    aics.append(gmm.aic(scaled_df))

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.plot(n_clusters_range, bics, marker='o')
plt.title('BIC - Bayesian Information Criterion')
plt.xlabel('Number of Clusters')
plt.ylabel('BIC')

plt.subplot(1, 2, 2)
plt.plot(n_clusters_range, aics, marker='o')
plt.title('AIC - Akaike Information Criterion')
plt.xlabel('Number of Clusters')
plt.ylabel('AIC')

plt.tight_layout()
plt.show()

optimal_clusters_bic = n_clusters_range[np.argmin(bics)]
optimal_clusters_aic = n_clusters_range[np.argmin(aics)]

print(f'Optimal number of clusters based on BIC: {optimal_clusters_bic}')
print(f'Optimal number of clusters based on AIC: {optimal_clusters_aic}')
```
